## 1. Simple decision: `if`

## 4. Three-way decision: `if` … `elif` … `else`
Change `x` to a positive number, a negative number, and `0`.

```python
x = 5

if x > 0:
    print("positive")
elif x < 0:
    print("negative")
else:
    print("zero")
```

Compare with the **nested** version. It does the same thing, but is harder to read:

```python
x = 5

if x > 0:
    print("positive")
else:
    if x < 0:
        print("negative")
    else:
        print("zero")
```

## 5. Multi-way decision: letter grade → score
Two ways to write the same program. Change `grade` and re-run.

```python
grade = "B"

if grade == "A":
    print(4)
elif grade == "B":
    print(3)
elif grade == "C":
    print(2)
elif grade == "D":
    print(1)
else:
    print(0)
```

`match-case` (Python 3.10+). The `_` is the default case. It only works for **exact values**, not ranges like `> 90`.

For more about `match-case` please refer to the `optional slide` in our slide deck

```python
grade = "B"

match grade:
    case "A":
        print(4)
    case "B":
        print(3)
    case "C":
        print(2)
    case "D":
        print(1)
    case _:
        print(0)
```

## 6. Boolean expressions
A condition is a **Boolean expression**: Python evaluates it to `True` or `False`.
Predict each result, then run the cell.

```python
print(5 > 3)
print(3 == 5)
print(len("hi") >= 2)
print(4 % 2 == 0)     # % gives the remainder after division
```

Python first **evaluates each side** of the operator, then **compares** the results.

| Read | Math | Python |
|---|:---:|:---:|
| Less than | < | `<` |
| Less than or equal to | ≤ | `<=` |
| Greater than | > | `>` |
| Greater than or equal to | ≥ | `>=` |
| Equal to | = | `==` |
| Not equal to | ≠ | `!=` |

Try your own comparisons below:

```python
x = "hi"
print(x == "hi")
print(x != "hi")
print(type(1 < 2))
```
