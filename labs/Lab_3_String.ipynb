## A3 - Accessing Characters (Indexing and Slicing)

A string is a **sequence of characters**. Each character has an index, and Python starts counting from `0`.

| Character | s | m | i | l | e |
|:---:|:---:|:---:|:---:|:---:|:---:|
| Index | 0 | 1 | 2 | 3 | 4 |
| Reverse index | -5 | -4 | -3 | -2 | -1 |

```
my_string[0]    # 's'
my_string[-1]   # 'e'
my_string[-2]   # 'l'
my_string[1:4]  # 'mil'
my_string[:3]   # 'smi'
```

## Part H — Searching, Counting, and Replacing

| Method | Purpose | Example | Result |
|---|---|---|---|
| `in` | Is the substring present? | `"He" in "Hello"` | `True` |
| `.index(x)` | Starting index of first match; **errors** if not found | `"Hello".index("l")` | `2` |
| `.find(x)` | Starting index of first match; returns **-1** if not found | `"Hello".find("z")` | `-1` |
| `.count(x)` | How many times `x` occurs | `"Hello".count("l")` | `2` |
| `.replace(a, b)` | Returns a new string with `a` replaced by `b` | `"Hello, World!".replace("World", "class")` | `"Hello, class!"` |

`.find()` is usually safer than `.index()` when you're not sure the substring is there, since it doesn't crash your program.


````python
sentence = "the cat sat on the mat"
print("cat" in sentence)
print(sentence.find("sat"))
print(sentence.count("the"))
print(sentence.replace("cat", "dog"))
````
