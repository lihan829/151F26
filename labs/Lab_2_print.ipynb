## Part A — `print()`: displaying output

`print()` is a **function**: you give it something inside the parentheses (called an *argument*), and it displays that value in the output area below the cell.

- Text (a *string*) must be inside quotes: `print("hello")`.
- Numbers can be printed directly, with no quotes: `print(5)`.
- You can print the value stored in a variable: `print(my_variable)`.
- You can print several things in one call by separating them with commas — `print()` automatically puts a space between them.
- Every call to `print()` starts on a new line in the output.

Run the examples below.


````python
print("Hello, linguists!")
print(5)
print(3 + 4)

language = "Tagalog"
print(language)
print("I am studying", language)
print("Two facts:", language, "and", 3 + 4)
````

## Part B — `input()`: collecting information

`input()` pauses your program, shows an optional prompt, and waits for the user to type something and press **Enter**. Whatever the user types is handed back to you **as a string**, even if it looks like a number.

```python
name = input("What is your name? ")
print("Hello,", name)
```

**Important:** because `input()` always returns a string, `input("How old are you? ")` gives you the text `"20"`, not the number `20`. If you want to calculate with it, convert it first, using `int()` for whole numbers or `float()` for numbers with a decimal point:

```python
age_text = input("How old are you? ")   # "20"  (a string)
age = int(age_text)                     # 20    (an integer)
print(age + 1)
```

You can also do the conversion in one line: `age = int(input("How old are you? "))`.


````python
name = input("What is your name? ")
print("Hello,", name)

age = int(input("How old are you? "))
print("Next year you will be", age + 1)
````

## Part C — Binary operators: doing arithmetic

A **binary operator** takes two values (its *operands*) and combines them into a single result — for example, `3 + 4` combines `3` and `4` with the `+` operator to produce `7`. Python has seven binary arithmetic operators:

| Operator | Name | Example | Result |
|:---:|---|:---:|:---:|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | division | `7 / 2` | `3.5` |
| `//` | floor (integer) division | `7 // 2` | `3` |
| `%` | modulus (remainder) | `7 % 2` | `1` |
| `**` | exponentiation | `7 ** 2` | `49` |

A few things worth noticing:

- `/` (true division) **always** returns a `float`, even when the division is exact: `8 / 2` is `4.0`, not `4`.
- `//` and `%` are a pair. `//` tells you how many whole times one number divides another; `%` tells you what is left over. So `(a // b) * b + (a % b)` always equals `a`.
- `**` is exponentiation (power), *not* the same as `^` in some other languages: `2 ** 5` means "2 to the power of 5" = `32`.
- Python follows the standard order of operations: `**` first, then `* / // %`, then `+ -`. Use parentheses `()` whenever you want to make the order explicit (and to make your code easier to read).
- `+` also works on strings — it **concatenates** (joins) them: `"Hello, " + "world!"`.


````python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)

# order of operations
print(2 + 3 * 4)      # multiplication happens first -> 14
print((2 + 3) * 4)    # parentheses force addition first -> 20

# + on strings concatenates instead of adding
print("Hello, " + "world!")
````
