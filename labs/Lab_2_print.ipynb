## Part 2. Variables

A **variable** gives a name to a value.

```python
word = "language"
```

Here:

- `word` is the **variable** name.
- `"language"` is a **string literal** — a value written directly in the code.
- The **type** of this value is `str` (string, wrapped in quotes).

Run the example below.

````python
word = "language"
print(word)
````

You can print multiple values in a single `print()` by separating them with commas.

## Part 4. Numbers from `input()`

Remember: `input()` always gives us a **string**, even when the user types a number.

Try this:


````python
age = input("How old are you? ")
print(age)
````

To use the answer as a number, convert it with `int()`:

```python
age = int(input("How old are you? "))
```

Now Python can do arithmetic with it.

````python
age = int(input("How old are you? "))
print("In the year of 2099 you will be", age + 2099 - 2024, "years old.")
````
