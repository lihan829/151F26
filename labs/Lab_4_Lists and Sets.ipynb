## A2 — Combining Lists: `+`, `*`, and `len()`

| Operator/Function | Meaning | Example | Result |
|:---:|---|---|---|

## A3 — Mutability: `.append()`, `.insert()`, `.remove()`, `.sort()`, `.reverse()`

- Recall that these are **methods** that apply to lists
- Unlike strings, lists are **mutable** meaning that these methods change the list *in place* and each one returns `None` (so never write `mylist = mylist.sort()`).

| Method | Effect |
|---|---|
| `.append(x)` | Add `x` to the end |
| `.insert(i, x)` | Add `x` at index `i` |
| `.remove(x)` | Remove the **first** occurrence of `x` |
| `.sort()` | Sort the list in place |
| `.reverse()` | Reverse the order in place |

## A5 — Strings ⇄ Lists: `.split()` and `.join()`

`.split()` breaks a string into a list of words (on whitespace by default, or on a separator you give it). `sep.join(list)` glues a list of strings back together, with `sep` inserted between items — notice it's called **on the separator**, not on the list.

```python
"the cat sat".split()             # ['the', 'cat', 'sat']
"2026-08-17".split('-')           # ['2026', '08', '17']
' '.join(['the', 'cat', 'sat'])   # 'the cat sat'
```

## Part D — Sets

A **set** is an unordered collection of *unique* items, written with curly braces `{ }` (or built from another sequence with `set(...)`). Duplicates are automatically dropped, which makes sets a quick way to find the **distinct** items in a list — like the vocabulary of a text.

```python
words_raw = ['dog', 'cat', 'dog', 'bird', 'cat']
print(set(words_raw))          # 3 unique words, order not guaranteed
print(len(words_raw), len(set(words_raw)))   # 5 tokens, 3 distinct words
```

Because a set is unordered, indexing (`myset[0]`) doesn't work, and **printing the same set twice in a row may show the items in a different order** — that's expected, not a bug. What you *can* rely on is membership (`in`), `.add(x)`, and the classic set operations:

| Operation | Meaning | Example |
|---|---|---|
| `a.union(b)` (or `a \| b`) | Items in `a` **or** `b` | all words used by either speaker |
| `a.intersection(b)` (or `a & b`) | Items in **both** `a` and `b` | words both speakers used |
| `a.difference(b)` (or `a - b`) | Items in `a` but **not** in `b` | words only the first speaker used |


````python
words_raw = ['dog', 'cat', 'dog', 'bird', 'cat']
print(set(words_raw))
print(len(words_raw), len(set(words_raw)))
````
